```python
# Problem 1: Smart Light Controller
# This class represents a smart light that can be turned ON or OFF.

class SmartLight:
    def __init__(self, light_name):
        # Constructor to initialize the light name
        self.light_name = light_name
        self.status = "OFF"   # Default status is OFF

    def turn_on(self):
        # Method to turn the light ON
        self.status = "ON"

    def turn_off(self):
        # Method to turn the light OFF
        self.status = "OFF"

    def display_status(self):
        # Method to display the current status of the light
        print(f"{self.light_name} is {self.status}")


# Creating object
light = SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

```python
# Problem 2: Employee ID Card System
# This class stores employee details and displays ID card information.

class Employee:
    def __init__(self, name, emp_id, department):
        # Initialize employee attributes
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        # Display employee ID card details
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# Creating object
emp = Employee("Rahul", "EMP102", "IT")
emp.display_id_card()
```

```python
# Problem 3: Mobile Contact Record
# This class stores contact details.

class Contact:
    def __init__(self, name, phone):
        # Initialize contact name and phone number
        self.name = name
        self.phone = phone

    def display_contact(self):
        # Display contact details
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Creating object
contact1 = Contact("Anita", "9876543210")
contact1.display_contact()
```

```python
# Problem 4: Product Price Tag Generator
# This class generates a formatted price tag for a product.

class Product:
    def __init__(self, product_name, price):
        # Initialize product details
        self.product_name = product_name
        self.price = price

    def display_price_tag(self):
        # Display formatted price tag
        print(f"Product: {self.product_name}")
        print(f"Price: ₹{self.price}")


# Creating object
product1 = Product("Headphones", 2499)
product1.display_price_tag()
```

```python
# Problem 5: Movie Rating Display System
# This class stores movie details and rating.

class Movie:
    def __init__(self, movie_name, rating):
        # Initialize movie attributes
        self.movie_name = movie_name
        self.rating = rating

    def display_movie_details(self):
        # Display movie information
        print(f"Movie: {self.movie_name}")
        print(f"Rating: {self.rating} / 5")


# Creating object
movie1 = Movie("Inception", 4.8)
movie1.display_movie_details()
```

```python
# Problem 6: Delivery Address Manager
# This class stores customer delivery details.

class Delivery:
    def __init__(self, customer_name, address):
        # Initialize delivery details
        self.customer_name = customer_name
        self.address = address

    def display_delivery_details(self):
        # Display delivery information
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# Creating object
delivery1 = Delivery("Suman", "Hyderabad")
delivery1.display_delivery_details()
```
